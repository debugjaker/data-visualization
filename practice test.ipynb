```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Pie plot
amount = [200, 130, 140, 150, 100]
name = ["emp1", "emp2", "emp3", "emp4", "emp5"]
explode = [0.2, 0, 0, 0.2, 0]
plt.pie(amount, labels = name, autopct="%1.1f%%", explode = explode, shadow = True)
```

```python
#Histogram plot

np.random.seed(0)
marks = np.random.normal(70, 10, 100)
plt.hist(marks, edgecolor = "black", color="red", bins = 5)
plt.title("Student & Marks")
plt.ylabel("# of Student")
plt.xlabel("Marks")
```

```python
#Stack plot

days = ["Mon", "Teu", "Wed", "Thu", "Fri", "Sat", "Sun"]
direct = [100, 120, 140, 160, 180, 200, 220]
organic = [40, 60, 70, 80, 90, 130, 150]
social = [115, 135, 145, 155, 165, 175, 185]
color = ["red", "blue", "green"]
plt.stackplot(days, direct, organic, social, labels = ["Direct", "Organic", "Social"], colors = color)
```

```python
#Box plot
N = [10, 12, 14, 16, 18, 20]
x1 = [np.sqrt(i) for i in N]
x2 = [i*2 for i in N]
x3 = [i**2 for i in N]
x4 = [i**3 for i in N]

plt.subplot(2, 2, 1)
plt.plot(N, x1)
plt.title("Plot - Root")

plt.subplot(2, 2, 2)
plt.plot(N, x2)
plt.title("Plot2 - double")

plt.subplot(2, 2, 3)
plt.plot(N, x3)
plt.title("Plot3 - Square")

plt.subplot(2, 2, 4)
plt.plot(N, x4)
plt.title("Plot - Cubes")
plt.tight_layout()
```
